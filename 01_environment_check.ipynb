# Environment check

This notebook reports the active Python interpreter and checks that the project's core data science packages can be imported. Run it with the project's `.venv` kernel to check that environment.

## Python interpreter

The following cell prints the Python version and exact interpreter path selected by the notebook kernel.

```python
import sys

print(f"Python version: {sys.version.split()[0]}")
print(f"Interpreter path: {sys.executable}")
```

## Package imports

Import NumPy, Pandas, scikit-learn, and XGBoost, then print each installed version. The final message appears only if every import succeeds.

```python
import numpy as np
import pandas as pd
import sklearn
import xgboost

print(f"NumPy version: {np.__version__}")
print(f"Pandas version: {pd.__version__}")
print(f"scikit-learn version: {sklearn.__version__}")
print(f"XGBoost version: {xgboost.__version__}")
print("Success: all required packages imported successfully.")
```
