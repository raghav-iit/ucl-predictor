# Data exploration

Explore the project dataset and summarize its structure, quality, and key patterns.

```python
from pathlib import Path
from urllib.request import urlopen

import pandas as pd
```

```python
PROJECT_ROOT = Path.cwd()
RAW_DIR = PROJECT_ROOT / "data" / "raw"
RAW_DIR.mkdir(parents=True, exist_ok=True)
```

```python
sources = {
    "2024-25": (
        "https://raw.githubusercontent.com/"
        "openfootball/champions-league/master/2024-25/cl.txt"
    ),
    "2025-26": (
        "https://raw.githubusercontent.com/"
        "openfootball/champions-league/master/2025-26/cl.txt"
    ),
}
```

```python
for season, url in sources.items():
    destination = RAW_DIR / f"ucl_{season}.txt"

    with urlopen(url, timeout=30) as response:
        content = response.read()

    destination.write_bytes(content)
    print(f"{season}: saved {destination}")
```

## 1. Inspecting the raw data

Before processing the historical UEFA Champions League match results, we inspect the raw files to understand their structure, the information they contain, and how individual matches are represented.

```python
from pathlib import Path

file_path = Path("data/raw/ucl_2024-25.txt")

raw_text = file_path.read_text()
lines = raw_text.splitlines()

print("Total lines:", len(lines))
print("\nFirst 30 lines:\n")

for line_number, line in enumerate(lines[:30], start=1):
    print(f"{line_number:>3}: {line}")
```
